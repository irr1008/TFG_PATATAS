##Analizar NVDI

NDVI_count representa el número de píxeles válidos utilizados en el cálculo
de las estadísticas del NDVI después de aplicar la máscara de nubes.

Se utiliza como indicador aproximado de la cantidad de información disponible
en cada fecha. No debe interpretarse como un porcentaje directo de nubosidad.

```python
df["NDVI_count"].describe()

ndvi_count_max = df["NDVI_count"].max()
quality_threshold = 0.5 * ndvi_count_max

print("Máximo NDVI_count:", ndvi_count_max)
print("Umbral orientativo:", quality_threshold)

valid_quality = (
    df["NDVI_count"] >= quality_threshold
)

print("Observaciones totales:", len(df))
print("Observaciones válidas:", valid_quality.sum())
print("Observaciones descartadas:", (~valid_quality).sum())
```

# Preparación de la serie temporal

Las observaciones de Sentinel-2 no tienen necesariamente una frecuencia diaria.
En esta sección se prepara una serie temporal diaria para utilizarla
posteriormente en el simulador.

Se distinguirán dos tipos de valores:

- Valores observados directamente en fechas con imágenes Sentinel-2.
- Valores estimados mediante interpolación temporal.

La interpolación no crea nuevas observaciones satelitales, sino valores
aproximados entre dos fechas disponibles.

```python
# RESUMEN DE LA EXPLORACIÓN DEL CSV

# Asegurar que el DataFrame está ordenado por fecha
df = (
    df.sort_values("date")
      .reset_index(drop=True)
)

# ANALIZAR OBSERVACIONES POR FECHA

# Contar cuántas filas existen para cada fecha
date_counts = (
    df["date"]
    .value_counts()
    .sort_index()
)


# Número de fechas distintas
number_of_unique_dates = (
    df["date"]
    .nunique()
)


# Número de fechas con más de una observación
dates_with_multiple_observations = (
    date_counts > 1
).sum()


# Número de filas pertenecientes a fechas con
# más de una observación
rows_in_multiple_observation_dates = (
    date_counts[
        date_counts > 1
    ].sum()
)

# ANALIZAR CALIDAD DE LAS OBSERVACIONES ORIGINALES

# Máximo número de píxeles válidos por fila
ndvicount_max = (
    df["NDVI_count"]
    .max()
)


# Umbral orientativo del 50 %
quality_threshold = (
    0.5 *
    ndvicount_max
)


# Filas que superan el umbral de calidad
valid_quality_rows = (
    df["NDVI_count"]
    >= quality_threshold
)


# Número de filas válidas antes de agrupar
valid_rows_count = (
    valid_quality_rows
    .sum()
)

# SELECCIONAR COLUMNAS Y AGRUPAR POR FECHA

# Seleccionar las columnas de medias de los índices
mean_columns = [
    column
    for column in df.columns
    if column.endswith("_mean")
]


columns_for_daily = (
    mean_columns.copy()
)


# Añadir NDVI_count para evaluar la calidad por fecha
if "NDVI_count" in df.columns:
    columns_for_daily.append(
        "NDVI_count"
    )


# Agrupar las dos observaciones de cada fecha
# calculando la media
daily_observed = (
    df
    .groupby("date")[columns_for_daily]
    .mean()
    .sort_index()
)


# APLICAR EL FILTRO DESPUÉS DE AGRUPAR

# Máximo NDVI_count después de agrupar
grouped_count_max = (
    daily_observed["NDVI_count"]
    .max()
)


# Umbral de calidad después de agrupar
grouped_quality_threshold = (
    0.5 *
    grouped_count_max
)


# Fechas que superan el umbral de calidad
valid_grouped_quality = (
    daily_observed["NDVI_count"]
    >= grouped_quality_threshold
)


# Número de fechas válidas después de agrupar
valid_dates_count = (
    valid_grouped_quality
    .sum()
)


# DataFrame agrupado y filtrado
daily_quality = (
    daily_observed[
        valid_grouped_quality
    ]
    .copy()
)

# CREAR EL RESUMEN

summary = pd.DataFrame({
    "Métricas": [
        "Número de filas originales",
        "Número de columnas",
        "Número de fechas distintas",
        "Primera fecha",
        "Última fecha",
        "Fechas con múltiples observaciones",
        "Filas pertenecientes a fechas múltiples",
        "Valores nulos en NDVI_mean",
        "NDVI_count máximo por fila",
        "Umbral de calidad por fila",
        "Filas válidas antes de agrupar",
        "Filas después de agrupar",
        "NDVI_count máximo agrupado",
        "Umbral de calidad agrupado",
        "Fechas válidas después de agrupar"
    ],
    "Valores": [
        len(df),
        df.shape[1],
        number_of_unique_dates,
        df["date"].min(),
        df["date"].max(),
        dates_with_multiple_observations,
        rows_in_multiple_observation_dates,
        df["NDVI_mean"].isna().sum(),
        ndvicount_max,
        quality_threshold,
        valid_rows_count,
        len(daily_observed),
        grouped_count_max,
        grouped_quality_threshold,
        valid_dates_count
    ]
})


# Mostrar el resumen
display(summary)

# GUARDAR EL RESUMEN

# Crear la carpeta de resultados
results_path = (
    "/content/drive/MyDrive/"
    "Sentinel2_Indices/"
    "resultados_exploracion"
)

os.makedirs(
    results_path,
    exist_ok=True
)


# Definir la ruta de salida
summary_path = os.path.join(
    results_path,
    "resumen_exploracion.csv"
)


# Guardar el resumen
summary.to_csv(
    summary_path,
    index=False,
    encoding="utf-8-sig"
)


print(
    "Resumen guardado correctamente en:"
)

print(summary_path)

# COMPROBACIONES FINALES

print("\nComprobaciones:")
print(
    "Filas originales:",
    len(df)
)

print(
    "Fechas distintas:",
    df["date"].nunique()
)

print(
    "Filas después de agrupar:",
    len(daily_observed)
)

print(
    "Fechas duplicadas después de agrupar:",
    daily_observed.index.duplicated().sum()
)

print(
    "Fechas conservadas después del filtro:",
    len(daily_quality)
)
```

- `NDVI_count` no se ha interpolado porque representa una medida de calidad
  asociada a imágenes reales.
- La serie diaria es una preparación para el simulador, no una colección de
